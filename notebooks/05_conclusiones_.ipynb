# Proyecto Integrador - Minería de Datos I

**Notebook 05: Conclusiones**


**Objetivo**

En este notebook se presentan las conclusiones finales del proyecto integrador a partir de los resultados obtenidos durante las etapas de inspección inicial, limpieza de datos, análisis exploratorio y reducción de dimensionalidad mediante PCA.

También se exponen las principales limitaciones del trabajo y posibles líneas de mejora para futuros análisis.

# Cumplimiento de los objetivos

Durante el desarrollo del proyecto se logró:

- Inspeccionar y comprender la estructura del conjunto de datos.
- Identificar problemas de calidad, como registros duplicados, valores faltantes, categorías inconsistentes, fechas inválidas y valores fuera de rango.
- Aplicar un proceso de limpieza y preparación de datos basado en evidencia.
- Realizar un análisis exploratorio mediante visualizaciones univariadas, bivariadas y multivariadas.
- Aplicar el Análisis de Componentes Principales (PCA) sobre las variables numéricas previamente escaladas.
- Documentar cada etapa del proceso para garantizar la reproducibilidad del proyecto.

# Hallazgos principales

El análisis exploratorio permitió identificar varios aspectos relevantes del comportamiento del conjunto de datos:

- El plan **Básico** es el más utilizado por los usuarios.
- La mayor parte del tiempo mensual de visualización se concentra entre aproximadamente **500 y 1000 minutos**.
- Los usuarios del plan **Premium** presentan, en general, un mayor tiempo de visualización que los usuarios de los demás planes.
- No se observó una relación lineal clara entre la edad y el tiempo mensual de visualización.
- Las variables numéricas presentan correlaciones bajas entre sí, indicando que cada una aporta información diferente al análisis.

# Resultados del Análisis de Componentes Principales

El PCA se aplicó sobre las variables numéricas previamente escaladas.

Los resultados mostraron que la varianza se distribuye de manera relativamente uniforme entre las tres componentes principales, sin que una de ellas concentre una proporción significativamente mayor de la información.

Este comportamiento confirma lo observado durante el análisis exploratorio, donde las variables presentaban correlaciones bajas. En consecuencia, la reducción de dimensionalidad resultó limitada, siendo necesario conservar las tres componentes para representar la totalidad de la información disponible.

# Limitaciones

El alcance de las conclusiones se encuentra condicionado por las características del conjunto de datos analizado.

Algunas variables categóricas no fueron utilizadas en el PCA debido a la naturaleza de la técnica, y la disponibilidad de información limita el análisis a las variables presentes en el dataset.

Asimismo, los resultados obtenidos corresponden únicamente al conjunto de datos proporcionado y no deben generalizarse a otros contextos sin realizar análisis adicionales.

# Mejoras futuras

Como trabajo futuro podrían incorporarse nuevas variables relacionadas con el comportamiento de los usuarios, como la frecuencia de acceso, el tipo de contenido consumido o el dispositivo utilizado.

También sería posible aplicar técnicas de clustering o modelos predictivos para identificar perfiles de usuarios y ampliar el alcance del análisis realizado en este proyecto.

# Conclusión final

El desarrollo del proyecto permitió aplicar de manera integrada las principales etapas del proceso de minería de datos, desde la inspección inicial hasta la reducción de dimensionalidad mediante PCA.

Las decisiones de limpieza y preparación fueron justificadas a partir de la evidencia observada en el conjunto de datos, permitiendo obtener un dataset consistente para el análisis.

Finalmente, el análisis exploratorio y el PCA permitieron comprender las características del conjunto de datos y responder a los objetivos planteados, garantizando un proceso reproducible y correctamente documentado.
